# EDA — Antimicrobial Resistance Dataset

```python
import pandas as pd
df = pd.read_csv('../data/culture_episodes.csv')
df.shape
```

```python
resist_cols = [c for c in df.columns if c.startswith('resistant_')]
df[resist_cols].mean().sort_values(ascending=False)
```

## Resistance prevalence drift over simulated months (antibiogram drift)

```python
df.groupby('month')['resistant_carbapenems'].mean().plot(title='Carbapenem resistance drift by month')
```
